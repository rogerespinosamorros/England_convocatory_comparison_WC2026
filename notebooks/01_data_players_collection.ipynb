```python
# Function to clean every dataframe from FBRef
def load_fbref_table(path: str) -> pd.DataFrame:
    """
    Load and clean a FBref table exported as CSV.

    FBref repeats table headers every 25 rows when exporting
    large tables. This function removes those duplicated
    header rows and returns a clean DataFrame.

    Args:
        path (str): Path to the FBref CSV file.

    Returns:
        pd.DataFrame: Cleaned FBref table.
    """
    df = pd.read_csv(path, header=1)

    return (
        df[df["Player"] != "Player"]
        .copy()
        .reset_index(drop=True)
    )
```
