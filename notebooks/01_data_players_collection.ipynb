```python
# Loading CSV
df = pd.read_csv("../data/raw/csv/standard_stats_top5_leagues.csv", header=1)
df.info() # All strings
```

```python
# Fixing the Minutes column, comas 
df["Min"] = pd.to_numeric(
    df["Min"].str.replace(",", "", regex=False),
    errors="coerce"
)
```

```python
# Checking type "Min"
df["Min"].dtype

df.info() # All strings, except for the Minutes column

```

```python
# df with english fotballers only, and copy to avoid SettingWithCopyWarning
eng = df[df["Nation"].str.contains("ENG", na=False)].copy()
# Checking type
print(eng["Min"].dtype)
```

```python
# values Minutes
eng["Min"].sample(20)
```

```python
eng.info() # All strings, except Minutes
```

```python
# Only english players with +800 minutes played
eng_800 = eng[eng["Min"] >= 800]
print(f"Number of English players: {len(eng)}")
print(f"Number of English players with more than 800 minutes played: {len(eng_800)}")
```

```python
# Saving English players with more than 800 minutes played in a CSV
eng_800.to_csv("../data/processed/eng_players_800min.csv", index=False)
```
